# <CENTER>  **FIGURE 1.**
## ====> Annilus of the form x² + y² = 1 
## ====> Application: Voltage
## ====> Boundary Condition: V = cos(3θ) (Dirichlet)

### **1.1** Laplace equation ∇²V(x) = 0

#### **Variables and data**

````julia
#variables for plotting 
df_voltage_no_source = CSV.read("./DATA/Voltage_annilus_laplace.csv", DataFrame)
x = df_voltage_no_source[:, "x position"]
y = df_voltage_no_source[:, "y position"]
sim = df_voltage_no_source[:, "Simulation Output (V)"]
analy = df_voltage_no_source[:, "Analytical Output (V)"]
rel_error = df_voltage_no_source[:, "Relative error (%)"]
avg_error = mean(rel_error)
println("average error: ϵ = ", avg_error)
show(df_voltage_no_source, allcols = true)
````

#### **Path animation for annilus**

````julia
pos = Brownian_motion_positions_laplace(0.0, 0.0, 100, exit_func_annilus)

anim = @animate for i in 1:10:length(pos)
    θ = 0:0.01:2π
    plot(cos.(θ), sin.(θ), label="Boundary", xlims=(-1.1, 1.1), ylims=(-1.1, 1.1), aspect_ratio=1,
        title="Brownian Motion Path", xlabel="X", ylabel="Y", legend=false, color=:red, lw=2)
    for j in 1:10:length(pos[i])
        plot!(map(p->p[1], pos[i][1:j]), map(p->p[2], pos[i][1:j]), seriestype=:scatter,
            legend=false, markersize=2)
        scatter!([pos[i][end][1]], [pos[i][end][2]], markersize=5, color=:gold)
    end
end
gif(anim, "./PLOTS/Annilus Laplace/Voltage_Brownian_path_laplace.gif", fps=1)
nothing
````

#### **Plots for analytical solution and simulation**

````julia

plt1 = scatter(x, y, marker_z=sim, 
    title="Laplace's Equation Monte Carlo Solution", 
    xlabel="x", ylabel="y", 
    color=:viridis, clims=(-1, 1), 
    aspect_ratio=1, legend=false, markersize=6)
    savefig("./PLOTS/Annilus Laplace/Simulated_Voltage_Laplace_2D.png")


plt2 = scatter(x, y, marker_z=analy, 
    title="Laplace's Equation Analytic Solution", 
    xlabel="x", ylabel="y", 
    color=:viridis, clims=(-1, 1),
    aspect_ratio=1, legend=false, markersize=6)
    savefig("./PLOTS/Annilus Laplace/Analytic_Voltage_Laplace_2D.png")

plt3 = scatter(x, y, sim, 
    title="Laplace's Equation Monte Carlo Solution", 
    xlabel="x", ylabel="y", zlabel = "Voltage",
    color=:viridis, clims=(-1, 1), legend=false, markersize=5)
    savefig("./PLOTS/Annilus Laplace/Simulated_Voltage_Laplace_3D.png")

plt4 = scatter(x, y, analy, 
    title="Laplace's Equation Analytic Solution", 
    xlabel="x", ylabel="y", zlabel = "Voltage",
    color=:viridis, clims=(-1, 1), legend=false, markersize=5)
    savefig("./PLOTS/Annilus Laplace/Analytic_Voltage_Laplace_3D.png")
    
    
plot(plt1, plt2, plt3, plt4, layout=(2, 2), size=(1000, 800), 
    xlabel="x", ylabel="y", zlabel = "Voltage", color=:viridis, clims=(-1, 1), legend=false, markersize=5)
    savefig("./PLOTS/Annilus Laplace/Voltage_Laplace_2D_3D.png")

plot!()
nothing
````

## ====> Flower of the form r(θ) = cos(2θ)
## ====> Application: Temperature
## ====> Boundary condition: T = 5*r(θ)

### **Variables and data**

````julia
df_data_flower = CSV.read("./DATA/Temp_data_flower.csv", DataFrame)

x = df_data_flower[:, "x position"]
y = df_data_flower[:, "y position"]
sim = df_data_flower[:, "Simulation Output (°C)"]
analy = df_data_flower[:, "Analytical Output (°C)"]
rel_error = df_data_flower[:, "Relative error (%)"]
avg_error = mean(rel_error)
println("average error: ϵ ≈ ", avg_error)
show(df_data_flower, allcols = true)
````

#### **Brownian path animation**

````julia
points_fl = Brownian_motion_positions_laplace(0.5, 0.0, 100, exit_func_flower)

anim = @animate for i in 1:10:length(points_fl)
    θ = collect(0:0.01:2π)
    r = cos.(2*θ)
    x_ = r.*cos.(θ)
    y_ = r.*sin.(θ)
    plot(x_, y_, label="Boundary", xlims=(-1.1, 1.1), ylims=(-1.1, 1.1), aspect_ratio=1,
        title="Brownian Motion Path", xlabel="X", ylabel="Y", legend=false, color=:red, lw=2)
    for j in 1:10:length(points_fl[i])
        plot!(map(p->p[1], points_fl[i][1:j]), map(p->p[2], points_fl[i][1:j]), seriestype=:scatter,
            legend=false, markersize=2)
        scatter!([points_fl[i][end][1]], [points_fl[i][end][2]], markersize=5, color=:gold)
    end
end
gif(anim, "./PLOTS/Flower/flower_Brownian_path_laplace.gif", fps=1)

nothing
````

#### **Plotting for analytical solution and simulation**

````julia
plt1 = scatter(x, y, marker_z=sim, 
    title="laplace's Equation Monte Carlo Solution", 
    xlabel="x", ylabel="y", 
    color=:thermal, clims=(-1, 1), 
    aspect_ratio=1, legend=false, markersize=6, colorbar=true)
    savefig("./PLOTS/Flower/Simulated_Temp_flower_2D.png")


plt2 = scatter(x, y, marker_z=analy, 
    title="laplace's Equation Analytic Solution", 
    xlabel="x", ylabel="y", 
    color=:thermal, clims=(-1, 1),
    aspect_ratio=1, legend=false, markersize=6, colorbar=true)
    savefig("./PLOTS/Flower/Analytic_Temp_flower_2D.png")
    
plt3 = scatter(x, y, sim, marker_z = sim,
    title="laplace's Equation Monte Carlo Solution", 
    xlabel="x", ylabel="y", zlabel = "Tempreature",
    color=:thermal, clims=(-1, 1), legend=false, markersize=5)
    savefig("./PLOTS/Flower/Simulated_Temp_flower_3D.png")

plt4 = scatter(x, y, analy, marker_z = analy,
    title="laplace's Equation Analytic Solution", 
    xlabel="x", ylabel="y", zlabel = "Tempreature",
    color=:thermal, clims=(-1, 1), legend=false, markersize=5)
    savefig("./PLOTS/Flower/Analytic_Temp_flower_3D.png")

    
plot(plt1, plt2, plt3, plt4, layout=(2, 2), size=(1000, 800))
    savefig("./PLOTS/Flower/Temp_flower_2D_3D.png")

plot!()
nothing
````
